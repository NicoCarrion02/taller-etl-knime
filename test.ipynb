```python
import pandas as pd
import os

# Ruta donde guardamos los CSV
ruta_carpeta = "./datasets/kaggle_f1/"

# Listar solo los archivos CSV del directorio
archivos_csv = [f for f in os.listdir(ruta_carpeta) if f.endswith('.csv')]

print("📊 Reporte de valores '\\N' por archivo y columna")
print("=" * 55)

for archivo in archivos_csv:
    ruta_completa = os.path.join(ruta_carpeta, archivo)
    
    try:
        # dtype=str garantiza que leamos el '\N' crudo antes de cualquier casteo automático
        df = pd.read_csv(ruta_completa, dtype=str)
        
        # Filtramos y contamos cuántas veces aparece exactamente el string '\N'
        conteos = (df == r'\N').sum()
        
        # Nos quedamos solo con las columnas que tienen > 0 valores encontrados
        columnas_con_n = conteos[conteos > 0]
        
        print(f"\n📁 {archivo} (Total de filas: {len(df):,})")
        print("-" * 35)
        
        if columnas_con_n.empty:
            print("  ✅ Limpio. No se encontraron valores '\\N'.")
        else:
            for columna, cantidad in columnas_con_n.items():
                porcentaje = (cantidad / len(df)) * 100
                print(f"  ⚠️ {columna}: {cantidad:,} valores ({porcentaje:.2f}%)")
                
    except Exception as e:
        print(f"\n❌ Error procesando el archivo {archivo}: {e}")
```
